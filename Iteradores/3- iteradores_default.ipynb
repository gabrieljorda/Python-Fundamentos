```python
import itertools
```

```python
# Range()
for i in range(5):
    print(i)
```

```python
# Iterar lista 
frutas = ["Maçã","Banana","Laranja"]
for fruta in frutas:
    print(fruta)
```

```python
#iterar Dicionario
pessoa = {"nome":"Alice",
          "idade":30}
for chave in pessoa:
    print(chave, pessoa[chave])
```

```python
#Enumerate()
for indice , fruta in enumerate(frutas):
    print(indice , fruta)
    
```

```python
# Zip
nomes = [ "Alice","Bob", "Carlos"]
idades = [30,25,35]
for nome , idade in zip(nomes , idades):
    print(nome, idade)
```

```python
# Itertools
for i in itertools.count(start=10, step=2):
    if i > 20:
        break
    print(i)
```
