```python
from typing import List
```

```python
class Estudantes:
    def __init__(self, nome :str , notas : list[float]) ->None:
        self.nome = nome 
        self.notas = notas
    
    def media (self) -> float:
        return sum(self.notas) / len(self.notas)
```

```python
def alunos_acima_da_media(alunos: List[Estudantes], media_minima:float) -> List[str]:
    return [ aluno.nome for aluno in alunos if aluno.media() >= media_minima]
```

```python
alunos = [ 
          Estudantes("Alice",[7.5,8.0,8.9]),
          Estudantes("Bob",[6.0,5.5,7.0]),
          Estudantes("Carlos",[8.5,9.0,10.0])
          ]
```

```python
resultados = alunos_acima_da_media(alunos, 8.0)
resultados
```
