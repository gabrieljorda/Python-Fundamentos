Walrus
- Esse operador é usado para atribuição de variáveis dentro de expressões
- É usado com o operador :=
- Permite que você atribua um valor a uma variável enquanto avalia a expressão

```python
numeros = [ 5,12,3,18,7,20]
```

```python
numeros_meiores_que_10 = [n for n in numeros if(maior := n ) > 10]
numeros_meiores_que_10
```

```python
linhas_processadas = []
with open("texto.txt","r",encoding="utf_8") as arquivo:
    for linha in arquivo:
        if(linha_limpa := linha.strip()):
            linhas_processadas.append(linha_limpa)
```
