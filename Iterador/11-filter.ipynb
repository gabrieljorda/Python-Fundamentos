Filter
-É utilizado para filtrar elementos de um iterável com base em uma função que pode reornar True ou false.
-O objetivo é permitir apenas os elementos que satisfazem uma condição.

```python
def e_par(x):
     return x% 2 == 0 
 
```

```python
numeros = [1,2,3,4,5,6]
```

```python
numeros_pares = filter(e_par , numeros)
list(numeros_pares)
```

```python
produtos = [
    {"nome":"Produto A","preco":10 , "em_estoque":True},
    {"nome":"Produto B","preco":20 , "em_estoque":False},
    {"nome":"Produto C","preco":30 , "em_estoque":True},
    {"nome":"Produto D","preco":40 , "em_estoque":False}
]
```

```python
def produto_em_estoque(produto):
    return produto["em_estoque"]
```

```python
produtos_disponiveis = filter(produto_em_estoque, produtos)
list(produtos_disponiveis)
```
