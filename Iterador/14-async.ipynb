```python
import aiohttp 
import asyncio
```

```python
async def tarefa(nome,tempo):
    print(f"Tarefa {nome} iniciada")
    await asyncio.sleep(tempo)
    print(f"Tarefa {nome} concluida")
```

```python
async def main():
    await asyncio.gather(
        tarefa("A", 2),
        tarefa("B", 1)
    )
```

```python
try:
    asyncio.get_running_loop()
except RuntimeError:
    asyncio.run(main())
```

```python
asyncio.ensure_future(main())
```
